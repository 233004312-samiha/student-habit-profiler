```python

# Check whether totals are approximately 24
approximately_24 = np.isclose(
    df["Total_Hours"],
    24,
    atol=1e-9
)

print("TOTAL HOURS VERIFICATION")
print("Total records:", len(df))
print("Approximately equal to 24:", approximately_24.sum())
print("Not approximately equal to 24:", (~approximately_24).sum())

# Examine deviations from 24
print("\nDEVIATION FROM 24 HOURS")
print((df["Total_Hours"] - 24).describe())

# Check totals after rounding
print("\nUNIQUE TOTALS AFTER ROUNDING TO 2 DECIMAL PLACES")
print(df["Total_Hours"].round(2).value_counts().sort_index())
```
